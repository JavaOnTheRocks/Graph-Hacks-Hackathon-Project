```python
import os

import requests
from dotenv import load_dotenv
```

```python
OPENWEATHER_API_KEY = os.getenv("OPENWEATHER_API_KEY")
WEATHER_UNITS = os.getenv("WEATHER_UNITS", "metric")
```

```python
if not OPENWEATHER_API_KEY:
    raise ValueError("OPENWEATHER_API_KEY is missing from .env")
```

```python
BASE_URL = "https://api.openweathermap.org/data/2.5"

session = requests.Session()

print("OpenWeather environment loaded successfully.")
print("Units:", WEATHER_UNITS)
print("API key configured:", bool(OPENWEATHER_API_KEY))
```

```python
response = requests.get(
    "https://api.openweathermap.org/data/2.5/weather",
    params={
        "q": "Jaipur,IN",
        "appid": OPENWEATHER_API_KEY,
        "units": WEATHER_UNITS,
    },
    timeout=20,
)

print("HTTP status:", response.status_code)
```

```python
if response.ok:
    data = response.json()

    print("\nSUCCESS: OpenWeather API key is working!")
    print("City:", data["name"])
    print("Temperature:", data["main"]["temp"], "°C")
    print("Humidity:", data["main"]["humidity"], "%")
    print("Conditions:", data["weather"][0]["description"])
else:
    print("API request failed:")
    try:
        print(response.json())
    except ValueError:
        print(response.text)
```
